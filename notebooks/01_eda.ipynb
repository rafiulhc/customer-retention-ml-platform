```python
print("duplicate rows:", df.duplicated().sum())
print("duplicate customer IDs:", df["customerID"].duplicated().sum())
```

```python
categorical = [c for c in df.columns if df[c].dtype != "float64" and df[c].nunique() < 10]
for col in categorical:
    print(f"\n{col}:", df[col].value_counts().to_dict())
```

```python
target = df["Churn"] == "Yes"
for col in [c for c in categorical if c != "Churn"]:
    rates = target.groupby(df[col]).mean().sort_values(ascending=False).round(3)
    print(f"\n{col}:", rates.to_dict())
```

```python
buckets = pd.cut(
    df["tenure"], bins=[-1, 6, 12, 24, 48, 72], labels=["0-6", "7-12", "13-24", "25-48", "49-72"]
)
target.groupby(buckets, observed=True).mean().round(3)
```

```python
estimate = df["tenure"] * df["MonthlyCharges"]
print(
    "correlation of TotalCharges with tenure × MonthlyCharges:", round(df["TotalCharges"].corr(estimate), 3)
)
```
